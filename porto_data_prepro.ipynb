```python
import pandas as pd
df = pd.read_csv(r'D:\Download\taxi+service+trajectory+prediction+challenge+ecml+pkdd+2015\train.csv')

```

```python
print(df.head())
print(df.columns)
```

```python
import ast

trajectory_porto = {}

trip_ids = []
start_times = []
trajectory_points = []
taxi_ids = []

for i in range(len(df)):
    
    one_poly = ast.literal_eval(df.loc[i, 'POLYLINE'])
    if len(one_poly) < 20:
        continue
    
    trip_ids.append(df.loc[i, 'TRIP_ID'])
    start_times.append(df.loc[i, 'TIMESTAMP'])
    trajectory_points.append(one_poly)
    taxi_ids.append(df.loc[i, 'TAXI_ID'])

trajectory_porto['trip_ids'] = trip_ids
trajectory_porto['start_times'] = start_times
trajectory_porto['trajectory_points'] = trajectory_points

trajectory_porto['taxi_ids'] = taxi_ids

new_df = pd.DataFrame({'trip_ids': trip_ids, 'start_times': start_times, 'taxi_ids': taxi_ids})
new_df.groupby('taxi_ids').size()
```

```python
import matplotlib.pyplot as plt
import numpy as np

all_lengths = [len(one_poly) for one_poly in trajectory_points]
print(f'Total number of trajectories: {len(all_lengths)}')

print(f'Minimum trajectory length: {min(all_lengths)}')
print(f'Maximum trajectory length: {max(all_lengths)}')
print(f'Average trajectory length: {sum(all_lengths)/len(all_lengths)}')

lengs = np.array(all_lengths)
print(f'95% quantile of trajectory length: {np.quantile(lengs, 0.95, method="inverted_cdf")}')

# plt.hist(all_lengths, bins=300)
# plt.xlabel('Trajectory Length')
# plt.ylabel('Frequency')
# plt.title('Distribution of Trajectory Lengths')
# plt.show()
```

```python
import json

trajectory_porto['taxi_ids'] = [int(one_id) for one_id in taxi_ids]
trajectory_porto['trip_ids'] = [str(one_id) for one_id in trip_ids]
trajectory_porto['start_times'] = [int(one_time) for one_time in start_times]

print(trajectory_porto['taxi_ids'])
f  = open('./porto_trajectory.json', 'w')
json.dump(trajectory_porto, f, ensure_ascii=False)
f.close()
```
